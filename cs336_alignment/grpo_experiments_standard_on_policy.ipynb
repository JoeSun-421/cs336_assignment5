```python
import os

from cs336_alignment.vllm_utils import generate_completions

base_url = os.environ["VLLM_BASE_URL"].rstrip("/")
model_id = os.environ["VLLM_MODEL_ID"]

results = generate_completions(
    vllm_base_url=base_url,
    model_id=model_id,
    prompts=["What is 2 + 2?"],
    sampling_params={
        "temperature": 0.7,
        "max_tokens": 64,
        "n": 1,
        "seed": 0,
    },
)

print(results[0].text)
```
